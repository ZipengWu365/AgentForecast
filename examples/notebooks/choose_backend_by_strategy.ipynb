# Choose backend by strategy

This notebook shows how to keep model choice hidden behind a small strategy vocabulary.

```python
from agentforecast.backends import route_backends
route_backends(history_len=120, horizon=30, strategy="fast", exogenous_cols=[])
```

Try different goals: fast, accurate, streaming, low_data, long_horizon.

```python
for strategy in ["fast", "accurate", "streaming", "low_data", "long_horizon"]:
    print(strategy, route_backends(history_len=120, horizon=30, strategy=strategy, exogenous_cols=[])["candidate_backends"])
```

The key idea: users and agents do not need to learn every backend. They start from intent, not estimator trivia.
